```python
import pandas as pd
import joblib
import matplotlib.pyplot as plt

# -------------------------------
# 1. Load Trained ML Model
# -------------------------------
model_path = "../models/kmrl_lightgbm.joblib"
clf = joblib.load(model_path)
print(f" LightGBM model loaded from {model_path}")

# -------------------------------
# 2. Load Optimization Output
# -------------------------------
schedule_path = "../data/final_schedule.csv"
final_schedule = pd.read_csv(schedule_path)
print(f" Final optimized schedule loaded from {schedule_path}")

print("\nSample of Final Schedule:")
print(final_schedule.head())

# -------------------------------
# 3. Visualization
# -------------------------------
plt.figure(figsize=(6,6))
final_schedule["OptimizedDecision"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Train Allocation (Service vs Standby vs IBL)")
plt.ylabel("")
plt.show()

# -------------------------------
# 4. Save Backup Copies (Optional)
# -------------------------------
final_schedule.to_csv("../data/final_schedule_backup.csv", index=False)
joblib.dump(clf, "../models/kmrl_lightgbm_backup.joblib")
print(" Backups saved successfully.")
```
