```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("../data/kmrl_enhanced_dataset.csv")
print(df.head())

# Summary of dataset
print("\n--- Info ---")
print(df.info())

# Distribution of Decision_Today
plt.figure(figsize=(6,4))
df['Decision_Today'].value_counts().plot(kind='bar')
plt.title("Decision Distribution")
plt.xlabel("Decision_Today")
plt.ylabel("Count")
plt.show()
```
